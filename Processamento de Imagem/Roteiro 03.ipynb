```python
import cv2
import matplotlib.pyplot as plt
from skimage.filters import gaussian
import numpy as np
```

```python
image = cv2.imread('forest.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
plt.imshow(image, cmap='gray')
plt.show()


fig, axes = plt.subplots(2, 2, figsize=(10,8)) #para plotar mais de uma imagem
axes[0,0].imshow(image, cmap='gray')
axes[0,1].imshow(image, cmap='gray')
axes[1,0].imshow(image, cmap='gray')
axes[1,1].imshow(image, cmap='gray')
plt.show()
```

```python
image1 = cv2.imread('pcb1.png', cv2.IMREAD_GRAYSCALE)

image_sobel_x = cv2.Sobel(image1, ddepth=-1, dx=1, dy=0, ksize=3)
image_sobel_y = cv2.Sobel(image1, ddepth=-1, dx=0, dy=1, ksize=3)


fig, axes = plt.subplots(1, 3, figsize=(10,8)) #para plotar mais de uma imagem
axes[0].imshow(image1, cmap='gray')
axes[1].imshow(image_sobel_x, cmap='gray')
axes[2].imshow(image_sobel_y, cmap='gray')

plt.show()
```

```python
from skimage.filters import sobel, sobel_h, sobel_v

image_sobel = sobel(image1)
image_sobel_h = sobel_h(image1)
image_sobel_v = sobel_v(image1)
fig, axes = plt.subplots(1, 3, figsize=(10,8)) #para plotar mais de uma imagem
axes[0].imshow(image_sobel, cmap='gray')
axes[1].imshow(image_sobel_h, cmap='gray')
axes[2].imshow(image_sobel_v, cmap='gray')

plt.show()
```

```python
from skimage.filters import prewitt, prewitt_h, prewitt_v

image_prewitt = prewitt(image1)
image_prewitt_h = prewitt_h(image1)
image_prewitt_v = prewitt_v(image1)
fig, axes = plt.subplots(1, 3, figsize=(10,8)) #para plotar mais de uma imagem
axes[0].imshow(image_prewitt, cmap='gray')
axes[1].imshow(image_prewitt_h, cmap='gray')
axes[2].imshow(image_prewitt_v, cmap='gray')

plt.show()
```

```python
image_lap = cv2.Laplacian(image1, ddepth=cv2.CV_64F, ksize=1)
image_lapl_a = np.uint8(np.absolute(image_lap)) #ou
image_lapl_a = cv2.Laplacian(image1, ddepth=-1, ksize=1)

fig, axes = plt.subplots(1, 2, figsize=(10,8)) #para plotar mais de uma imagem
axes[0].imshow(image_lap, cmap='gray')
axes[1].imshow(image_lapl_a, cmap='gray')


plt.show()
```

```python
image2 = cv2.imread('pcb_cap.png', cv2.IMREAD_GRAYSCALE)

image_lap = cv2.Laplacian(image2, ddepth=cv2.CV_64F, ksize=1)
image_lapl_a = np.uint8(np.absolute(image_lap)) #ou
image_lapl_b = cv2.Laplacian(image2, ddepth=-1, ksize=1)

fig, axes = plt.subplots(1, 3, figsize=(10,8)) #para plotar mais de uma imagem
axes[0].imshow(image_lap, cmap='gray')
axes[1].imshow(image_lapl_a, cmap='gray')
axes[2].imshow(image_lapl_b, cmap='gray')

plt.show()
```

```python
image3 = cv2.imread('letters.jpg', cv2.IMREAD_GRAYSCALE)

image_canny = cv2.Canny(image3, 800/6, 800, apertureSize=3)


fig, axes = plt.subplots(1, 2, figsize=(10,8)) #para plotar mais de uma imagem
axes[0].imshow(image3, cmap='gray')
axes[1].imshow(image_canny, cmap='gray')
plt.show()
```

```python
image4 = cv2.imread('fingerprint.png', cv2.IMREAD_GRAYSCALE)
```

```python
hist_image = cv2.calcHist([image4], [0], None, [256], [0, 255])
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10,5)) #para plotar mais de uma imagem
axes[0].imshow(image4, cmap='gray')
axes[1].plot(hist_image)
plt.show()
```

```python
(T, image_bin) = cv2.threshold(image4, 10, 255, cv2.THRESH_BINARY)
hist_image1 = cv2.calcHist([image_bin], [0], None, [256], [0, 256])
fig, axes = plt.subplots(1, 2, figsize=(10,5)) #para plotar mais de uma imagem
axes[0].imshow(image_bin, cmap='gray')
axes[1].plot(hist_image1)
plt.show()
```

```python
image7 = cv2.imread('levedura.tif', cv2.IMREAD_GRAYSCALE)
```

```python
image_histo = cv2.calcHist([image7], [0], None, [256], [0,255])
(T, image_binario_seg) = cv2.threshold(image7, 0, 255, cv2.THRESH_BINARY+cv2.THRESH_OTSU)
print(T)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(20,5))
axes[0].imshow(image7, cmap='gray')
axes[1].imshow(image_binario_seg, cmap='gray')
axes[2].plot(image_histo)
plt.tight_layout
plt.show()
```

```python
image_lap_seg = cv2.Laplacian(image7, ddepth=-1, ksize=3)
(T, image_binario_seg) = cv2.threshold(image_lap_seg, 0, 255, cv2.THRESH_BINARY+cv2.THRESH_OTSU)
print(T)
```

```python
image_binario_seg = np.uint8(np.absolute(image_binario_seg))

imm = image_lap_seg * 2 * image7
```

```python
image_histo_lap = cv2.calcHist([imm], [0], None, [256], [1,255])
(T, image_binario_seg_1) = cv2.threshold(imm, 0, 255, cv2.THRESH_BINARY+cv2.THRESH_OTSU)
(T, image_binario_seg_1) = cv2.threshold(image7, T, 255, cv2.THRESH_BINARY)
print(T)
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20,5))
axes[0].plot(image_histo_lap)
axes[1].imshow(image_lap_seg, cmap='gray')
axes[2].imshow(imm, cmap='gray')
axes[3].imshow(image_binario_seg_1, cmap='gray')
plt.tight_layout
plt.show()
```
